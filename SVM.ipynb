```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv('used_car_regression_dataset.csv')
```

```python
X=data.iloc[:,:-1]
y=data.iloc[:,-1]
print(X)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
```

```python
numeric_features=['Year','Kilometers_Driven','Engine_CC','Mileage_Kmpl']
categorical_features=['Brand','Fuel_Type','Transmission','Owner']
```

```python
numeric= Pipeline([
    ('num',SimpleImputer(strategy='mean')),
    ('cat',StandardScaler())
])
category = Pipeline([
    ('imp',SimpleImputer(strategy='most_frequent')),
                              ('one',OneHotEncoder(handle_unknown='ignore'))])
```

```python
process=ColumnTransformer([
    ('num',numeric,numeric_features),
    ('cat',category,categorical_features)
])
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=42)
```

```python
from sklearn.svm import SVR
model = Pipeline([
    ("process", process),
    ("model", SVR())
])
```

```python
model.fit(X_train,y_train)
```

```python
pred=model.predict(X_test)

print(np.concatenate(y_test.to_numpy.reshape(-1,1),pred.to_numpy.reshape(-1,1)),1)
```

```python
from sklearn.metrics import r2_score
r2 =r2_score(y_test,pred) 
```
